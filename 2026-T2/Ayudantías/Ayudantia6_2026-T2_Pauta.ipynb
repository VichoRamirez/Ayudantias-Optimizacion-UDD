## Sección 0 — Resumen Teórico

En las clases anteriores dejamos el modelamiento y pasamos al **motor** que resuelve los programas lineales: el **Método Simplex**. En esta ayudantía repasamos las **dos caras del mismo algoritmo** —la forma **tableau** (operaciones de fila) y la forma **matricial** (con $B^{-1}$)— y mostramos explícitamente **cómo se arma el tableau desde las matrices**. Luego construimos el **problema dual**, que nos entrega los **precios sombra** y las **condiciones de holgura complementaria**.

Los tres bloques se explican por separado y luego se cruzan en los ejercicios.

### Convención: el Simplex se plantea como **minimización**

La forma canónica sobre la que trabaja el Simplex es siempre de **minimización**:

$$\min\; w=c^\top x \qquad \text{s.a.}\quad Ax\le b,\quad x\ge 0.$$

Un problema de **maximización** se resuelve **convirtiéndolo a mínimo**: maximizar es lo mismo que minimizar el negativo,

$$\max\; z=p^\top x \qquad\Longleftrightarrow\qquad \min\; w=-p^\top x,\qquad\text{con}\quad z=-w.$$

En la práctica esto significa una sola cosa: los **coeficientes de costo entran con el signo cambiado** ($c_j=-p_j$). De ahí en adelante **el tableau y todas las operaciones son idénticas**; al final se recupera el valor del máximo con $z^*=-w^*$. Haremos **todos los ejercicios así**, tal como la forma general de las clases ($\min c\cdot x$).

### Tema 1 — Método Simplex: forma **Tableau**

Trabajamos ya en forma de **minimización** $\min\,w=c^\top x$, con restricciones $\le$ y $b\ge 0$ (si el problema era un máximo, $c$ ya viene con el signo cambiado).

**Paso 1 — Forma estándar.** A cada restricción $\le$ se le suma una **variable de holgura** $s_i\ge 0$:

$$a_{i1}x_1+\dots+a_{in}x_n+s_i=b_i.$$

Las holguras forman de inmediato una **base factible inicial** ($B=I$, con $x_B=b\ge 0$).

**Paso 2 — Tableau.** Se ordenan los coeficientes en una tabla. La fila superior es la de **costos reducidos** (CR); las demás filas son las restricciones, etiquetadas por la variable básica. En la celda objetivo llevamos el valor $z=-w$ del problema original.

**Paso 3 — Criterio de entrada.** El **costo reducido** de la variable $j$ es $\;\bar c_j=c_j-z_j$, con $z_j=c_B B^{-1}A_j$. Para el problema de **minimización**:

- Si **todos** los $\bar c_j\ge 0$ → la solución es **óptima** (detenerse).
- Si hay $\bar c_j<0$ → **entra a la base** la variable con el costo reducido **más negativo** (es la que más hace bajar $w$).

> Como partimos de un máximo con $c_j=-p_j$, la fila CR inicial es $[\,-p_1,\dots,-p_n\,]$: **los márgenes con el signo cambiado**. Ese es el único efecto de la conversión; el resto del tableau queda exactamente igual.

**Paso 4 — Prueba del cociente (variable que sale).** Sea $x_k$ la que entra. Se calcula, **solo** para las filas con coeficiente positivo en la columna $k$,

$$\theta=\min_{i:\,a_{ik}>0}\frac{b_i}{a_{ik}},$$

y sale de la base la variable de la fila donde se alcanza el mínimo. El coeficiente $a_{ik}$ de esa fila es el **pivote**.

**Paso 5 — Pivoteo (Gauss-Jordan).** Se normaliza la fila pivote (dividir por el pivote) y se hace cero la columna $k$ en las demás filas, incluida la fila CR. Se vuelve al Paso 3.

### Tema 2 — Método Simplex: forma **Matricial** (y el cruce con el tableau)

La forma matricial hace **exactamente las mismas iteraciones**, pero en lugar de arrastrar toda la tabla, recalcula $B^{-1}$ y reconstruye lo que necesita. Dada una base $B$ (submatriz de columnas básicas) con costos $c_B$, y llamando $A_{NB}$ a las columnas no básicas:

| Fórmula | Qué es en el tableau |
|:---|:---|
| $B^{-1}b$ | la **columna RHS** (valores de las variables básicas $x_B$) |
| $B^{-1}A_j$ | la **columna** actualizada de la variable $j$ |
| $c_{NB}-c_B B^{-1}A_{NB}$ | la **fila de costos reducidos** de las no básicas |
| $w=c_B B^{-1}b$ | el **valor del mínimo** ($z=-w$ para un máximo convertido) |

**El cruce clave.** Como las columnas originales de las **holguras** forman la identidad $I$, ocurre que en cualquier iteración:

$$\underbrace{B^{-1}[\,A\mid I\mid b\,]}_{\text{tableau completo}}=[\,B^{-1}A\;\mid\; B^{-1}\;\mid\; B^{-1}b\,].$$

Es decir:

- Las **columnas bajo las holguras iniciales** $s_1,\dots,s_m$ del tableau **son literalmente $B^{-1}$**.
- La **columna RHS** es $B^{-1}b=x_B$.
- La **fila CR bajo la variable $j$** es $c_j-c_B B^{-1}A_j$.
- La **fila CR bajo las holguras** es $-c_B B^{-1}$: **esos son los precios sombra** $y^\top$ (del problema original).

Por eso "armar el tableau" y "calcular con $B^{-1}$" son la misma cosa: basta obtener $B^{-1}$ y multiplicar. Lo verificaremos número a número en el Ejercicio 1.

### Tema 3 — **Dualidad**

A todo programa lineal (**primal**) le corresponde un **dual**. Reglas de construcción (caso primal de **maximización** con $\le$):

| Primal ($\max\,c^\top x$) | Dual ($\min\,b^\top y$) |
|:---|:---|
| Restricción $i$ de tipo $\le b_i$ | Variable $y_i\ge 0$ |
| Variable $x_j\ge 0$ | Restricción $j$ de tipo $\ge c_j$ |
| $n$ variables, $m$ restricciones | $m$ variables, $n$ restricciones |

$$\boxed{\;\max\; c^\top x,\; Ax\le b,\; x\ge0 \qquad\Longleftrightarrow\qquad \min\; b^\top y,\; A^\top y\ge c,\; y\ge0\;}$$

**Dualidad débil.** Para todo par factible $(x,y)$: $\;c^\top x\le b^\top y$. (El dual acota al primal.)

**Dualidad fuerte.** Si el primal tiene óptimo $x^*$, el dual tiene óptimo $y^*$ y **coinciden**: $\;c^\top x^*=b^\top y^*$.

**Precios sombra.** Las variables duales son los **precios sombra** de las restricciones:
$$y_i^*=\frac{\partial z^*}{\partial b_i},$$
el valor marginal de relajar la restricción $i$ en una unidad. En el tableau óptimo se **leen directamente** en la fila CR bajo las holguras (al minimizar $-z$, corresponden a $-c_B B^{-1}$).

**Holguras complementarias** (equivalen a la condición KKT $\mu_i g_i(x^*)=0$). En el óptimo, para todo $i,j$:

$$y_i^*\,\big(b_i-a_i^\top x^*\big)=0,\qquad x_j^*\,\big(a_j^\top y^*-c_j\big)=0.$$

En palabras:
- Restricción primal **inactiva** (con holgura) $\Rightarrow$ su precio sombra es $y_i^*=0$.
- $y_i^*>0 \Rightarrow$ la restricción primal está **activa**.
- $x_j^*>0 \Rightarrow$ la restricción dual $j$ está **activa**.

## Ejercicio 1 — Simplex Tableau **y** Matricial

### Contexto de negocio

La **Cervecería artesanal "Los Andes"** produce dos estilos por lote semanal: **Golden Ale** ($x_1$, en barriles) y **Doble IPA** ($x_2$, en barriles). Los márgenes de contribución son **\$30 mil/barril** (Golden) y **\$50 mil/barril** (IPA). La producción semanal está limitada por tres recursos:

| Recurso | Golden ($x_1$) | IPA ($x_2$) | Disponible |
|:---|:---:|:---:|:---:|
| Malta (sacos) | $1$ | $2$ | $\le 16$ |
| Fermentación (días-tanque) | $1$ | $1$ | $\le 9$ |
| Embotellado (horas) | $3$ | $2$ | $\le 24$ |
| **Margen (\$ mil/barril)** | **30** | **50** | — |

**Se pide:** (a) plantear el modelo, llevarlo a **minimización** y a forma estándar; (b) resolver por **Simplex Tableau**; (c) resolver el **mismo** problema por **Simplex Matricial**, mostrando el cruce $B^{-1}b$, $B^{-1}A$, $c_{NB}-c_B B^{-1}A_{NB}$; (d) verificar con código.

### (a) Modelo y forma estándar

$$\max\; z=30x_1+50x_2 \qquad \text{s.a.}\quad
\begin{cases}
x_1+2x_2\le 16 & \text{(malta)}\\
x_1+x_2\le 9 & \text{(fermentación)}\\
3x_1+2x_2\le 24 & \text{(embotellado)}\\
x_1,x_2\ge 0
\end{cases}$$

**Conversión a mínimo.** Como el Simplex trabaja en forma de minimización, resolvemos $\min\,w=-z$, negando los márgenes:

$$\min\; w=-30x_1-50x_2,\qquad z^*=-w^*.$$

Agregando holguras $s_1,s_2,s_3\ge0$:

$$\begin{cases}
x_1+2x_2+s_1=16\\
x_1+x_2+s_2=9\\
3x_1+2x_2+s_3=24
\end{cases}
\qquad \min\; w=-30x_1-50x_2+0s_1+0s_2+0s_3.$$

Base inicial: $\{s_1,s_2,s_3\}$, con $B=I$. El vector de costos es $c=(-30,-50,0,0,0)$.

### (b) Resolución por Simplex Tableau

**Tableau 0** — Base $\{s_1,s_2,s_3\}$.  Fila CR $=\bar c_j=c_j-z_j$; al inicio ($z_j=0$) es el vector de costos $c=(-30,-50,0,0,0)$, es decir **los márgenes con signo cambiado**. La celda objetivo lleva $z=-w$.

| Base | $x_1$ | $x_2$ | $s_1$ | $s_2$ | $s_3$ | RHS |
|:---:|:---:|:---:|:---:|:---:|:---:|:---:|
| **CR** | $-30$ | $\mathbf{-50}$ | $0$ | $0$ | $0$ | $0$ |
| $s_1$ | $1$ | $2$ | $1$ | $0$ | $0$ | $16$ |
| $s_2$ | $1$ | $1$ | $0$ | $1$ | $0$ | $9$ |
| $s_3$ | $3$ | $2$ | $0$ | $0$ | $1$ | $24$ |

Más negativo: $-50$ → **entra $x_2$**. Cocientes: $16/2=8$, $9/1=9$, $24/2=12$ → mínimo $8$ → **sale $s_1$** (pivote $=2$).

**Tableau 1** — Base $\{x_2,s_2,s_3\}$. ($R_{x_2}=R_{s_1}/2$; $R_{s_2}\!-\!1\cdot R_{x_2}$; $R_{s_3}\!-\!2\cdot R_{x_2}$; $R_{CR}\!+\!50\,R_{x_2}$.)

| Base | $x_1$ | $x_2$ | $s_1$ | $s_2$ | $s_3$ | RHS |
|:---:|:---:|:---:|:---:|:---:|:---:|:---:|
| **CR** | $\mathbf{-5}$ | $0$ | $25$ | $0$ | $0$ | $400$ |
| $x_2$ | $1/2$ | $1$ | $1/2$ | $0$ | $0$ | $8$ |
| $s_2$ | $1/2$ | $0$ | $-1/2$ | $1$ | $0$ | $1$ |
| $s_3$ | $2$ | $0$ | $-1$ | $0$ | $1$ | $8$ |

Más negativo: $-5$ → **entra $x_1$**. Cocientes (columna $x_1$): $8/\tfrac12=16$, $1/\tfrac12=2$, $8/2=4$ → mínimo $2$ → **sale $s_2$** (pivote $=\tfrac12$).

**Tableau 2** — Base $\{x_2,x_1,s_3\}$. ($R_{x_1}=R_{s_2}/\tfrac12$; $R_{x_2}\!-\!\tfrac12R_{x_1}$; $R_{s_3}\!-\!2R_{x_1}$; $R_{CR}\!+\!5R_{x_1}$.)

| Base | $x_1$ | $x_2$ | $s_1$ | $s_2$ | $s_3$ | RHS |
|:---:|:---:|:---:|:---:|:---:|:---:|:---:|
| **CR** | $0$ | $0$ | $20$ | $10$ | $0$ | $\mathbf{410}$ |
| $x_2$ | $0$ | $1$ | $1$ | $-1$ | $0$ | $7$ |
| $x_1$ | $1$ | $0$ | $-1$ | $2$ | $0$ | $2$ |
| $s_3$ | $0$ | $0$ | $1$ | $-4$ | $1$ | $4$ |

Todos los CR $\ge 0$ → **óptimo**. La celda objetivo llegó a $-w=410$, es decir $w^*=-410$ y $z^*=-w^*=410$.

$$\boxed{x_1^*=2\text{ barriles Golden},\quad x_2^*=7\text{ barriles IPA},\quad z^*=-w^*=\$410\text{ mil}}$$

La restricción de **embotellado** queda con holgura $s_3=4$ (no activa); malta y fermentación están activas.

### (c) Resolución por Simplex Matricial

Columnas del sistema $[A\mid I]$ (orden $x_1,x_2,s_1,s_2,s_3$):

$$a_{x_1}=\begin{bmatrix}1\\1\\3\end{bmatrix},\;
a_{x_2}=\begin{bmatrix}2\\1\\2\end{bmatrix},\;
a_{s_1}=\begin{bmatrix}1\\0\\0\end{bmatrix},\;
a_{s_2}=\begin{bmatrix}0\\1\\0\end{bmatrix},\;
a_{s_3}=\begin{bmatrix}0\\0\\1\end{bmatrix},\qquad b=\begin{bmatrix}16\\9\\24\end{bmatrix}.$$

**Iteración 0** — Base $\{s_1,s_2,s_3\}$: $B=I$, $B^{-1}=I$, $c_B=[0,0,0]$.

$$x_B=B^{-1}b=[16,\,9,\,24]^\top,\qquad w=c_BB^{-1}b=0.$$
$$c_{NB}-c_BB^{-1}A_{NB}=[-30,-50]-[0,0,0]\!\cdot\![a_{x_1},a_{x_2}]=[-30,-50].$$
Más negativo $\Rightarrow$ entra $x_2$; cocientes $16/2,9/1,24/2 \Rightarrow$ sale $s_1$.

**Iteración 1** — Base $\{x_2,s_2,s_3\}$:

$$B=\begin{bmatrix}2&0&0\\1&1&0\\2&0&1\end{bmatrix},\quad
B^{-1}=\begin{bmatrix}\tfrac12&0&0\\[2pt]-\tfrac12&1&0\\[2pt]-1&0&1\end{bmatrix},\quad c_B=[-50,0,0].$$

$$x_B=B^{-1}b=\begin{bmatrix}8\\1\\8\end{bmatrix},\qquad w=c_BB^{-1}b=-400\;(z=400).$$

No básicas $\{x_1,s_1\}$:  $\;c_{NB}-c_BB^{-1}A_{NB}=[-5,\;25]$. Más negativo $-5\Rightarrow$ entra $x_1$.
Columna actualizada $B^{-1}a_{x_1}=[\tfrac12,\tfrac12,2]^\top$; cocientes $8/\tfrac12,1/\tfrac12,8/2\Rightarrow$ sale $s_2$.

**Iteración 2** — Base $\{x_2,x_1,s_3\}$:

$$B=\begin{bmatrix}2&1&0\\1&1&0\\2&3&1\end{bmatrix},\quad
B^{-1}=\begin{bmatrix}1&-1&0\\-1&2&0\\1&-4&1\end{bmatrix},\quad c_B=[-50,-30,0].$$

$$x_B=B^{-1}b=\begin{bmatrix}7\\2\\4\end{bmatrix}\;\Rightarrow\; x_2=7,\;x_1=2,\;s_3=4,\qquad w=c_BB^{-1}b=-410\;(z=410).$$

No básicas $\{s_1,s_2\}$: $\;c_{NB}-c_BB^{-1}A_{NB}=[20,\;10]\ge 0 \Rightarrow$ **óptimo**.

**El cruce con el tableau.** En el Tableau 2:
- las columnas bajo $s_1,s_2,s_3$ son $\begin{bmatrix}1&-1&0\\-1&2&0\\1&-4&1\end{bmatrix}=B^{-1}$;
- la columna RHS $[7,2,4]=B^{-1}b$;
- la fila CR bajo $s_1,s_2$ es $[20,10]=c_{NB}-c_BB^{-1}A_{NB}$, y bajo las tres holguras $-c_BB^{-1}=[20,10,0]$ son los **precios sombra** $y^\top$.

$$\boxed{x_1^*=2,\quad x_2^*=7,\quad z^*=-w^*=410,\qquad y^\top=(20,\,10,\,0)}$$

### (d) Comprobación con código

Iteramos el Simplex **matricial** en `numpy` (recalculando $B^{-1}$) y verificamos el óptimo con `scipy` y `gurobipy`. *(El código no es el foco: sirve solo de comprobación de lo hecho a mano.)*

```python
import numpy as np
from numpy.linalg import inv

# Columnas del sistema completo [A | I]  (variables x1, x2, s1, s2, s3)
cols = {'x1': np.array([1., 1., 3.]), 'x2': np.array([2., 1., 2.]),
        's1': np.array([1., 0., 0.]), 's2': np.array([0., 1., 0.]),
        's3': np.array([0., 0., 1.])}
# Forma de MINIMIZACION: min w = -30 x1 - 50 x2  (costos con signo cambiado)
c_obj = {'x1': -30, 'x2': -50, 's1': 0, 's2': 0, 's3': 0}
b = np.array([16., 9., 24.])
todas = ['x1', 'x2', 's1', 's2', 's3']

def iteracion(base):
    nb   = [v for v in todas if v not in base]
    B    = np.column_stack([cols[v] for v in base])
    Binv = inv(B)
    cB   = np.array([c_obj[v] for v in base], float)
    A_nb = np.column_stack([cols[v] for v in nb])
    c_nb = np.array([c_obj[v] for v in nb], float)
    red  = c_nb - cB @ Binv @ A_nb     # costos reducidos  c_j - z_j  (min)
    xB   = Binv @ b                     # B^-1 b  = valores basicos
    w    = cB @ Binv @ b               # valor del min ; z = -w
    print(f"Base {base} | NoBase {nb}")
    print("  B^-1 =", np.round(Binv, 4).tolist())
    print("  B^-1 b (x_B) =", np.round(xB, 4).tolist(), f" | w = {w:.2f}  (z = -w = {-w:.2f})")
    print(f"  costos reducidos {nb} = {np.round(red, 4).tolist()}")
    if np.all(red >= -1e-9):
        print("  >>> todos >= 0  =>  OPTIMO ;  precios sombra -cB*Binv =",
              np.round(-cB @ Binv, 2).tolist(), "\n")
    else:
        entra = nb[int(np.argmin(red))]
        colk  = Binv @ cols[entra]
        sale  = min((xB[i]/colk[i], base[i]) for i in range(len(base)) if colk[i] > 1e-9)[1]
        print(f"  entra {entra} (mas negativo), sale {sale}\n")

for base in [['s1','s2','s3'], ['x2','s2','s3'], ['x2','x1','s3']]:
    iteracion(base)
```

```python
# Verificacion del optimo con scipy y gurobipy
from scipy.optimize import linprog
res = linprog(c=[-30, -50], A_ub=[[1,2],[1,1],[3,2]], b_ub=[16,9,24],
              bounds=[(0,None),(0,None)], method='highs')
print("scipy : x1=%.2f  x2=%.2f  z*=%.2f" % (res.x[0], res.x[1], -res.fun))

import gurobipy as gp
from gurobipy import GRB
m = gp.Model("cerveceria"); m.Params.OutputFlag = 0
x1 = m.addVar(name="x1_golden"); x2 = m.addVar(name="x2_ipa")
m.setObjective(30*x1 + 50*x2, GRB.MAXIMIZE)
r1 = m.addConstr(x1 + 2*x2 <= 16, "malta")
r2 = m.addConstr(x1 +   x2 <=  9, "fermentacion")
r3 = m.addConstr(3*x1 + 2*x2 <= 24, "embotellado")
m.optimize()
print("gurobi: x1=%.2f  x2=%.2f  z*=%.2f" % (x1.X, x2.X, m.ObjVal))
for r in (r1, r2, r3):
    print(f"   {r.ConstrName:13s} precio sombra y={r.Pi:5.2f} | holgura={r.Slack:.2f}")
```

## Ejercicio 2 — Dualidad y Holguras Complementarias

### Contexto de negocio

El **Molino "Del Sur"** produce dos tipos de harina a partir de su stock semanal de trigo: **Harina integral** ($x_1$, en quintales) y **Harina blanca** ($x_2$, en quintales), con márgenes de **\$4** y **\$3** (mill\$/quintal). Las limitaciones semanales son:

| Recurso / condición | Integral ($x_1$) | Blanca ($x_2$) | Límite |
|:---|:---:|:---:|:---:|
| Trigo (toneladas) | $2$ | $1$ | $\le 10$ |
| Horas de molienda | $1$ | $1$ | $\le 7$ |
| Cuota máx. de integral | $1$ | $0$ | $\le 4$ |
| **Margen (mill\$/quintal)** | **4** | **3** | — |

**Se pide:** (a) resolver el primal (se entrega el tableau óptimo); (b) plantear el **dual**; (c) verificar **dualidad fuerte**; (d) verificar **holguras complementarias** e interpretar los **precios sombra**.

### (a) Primal y tableau óptimo

$$\max\; z=4x_1+3x_2 \qquad \text{s.a.}\quad
\begin{cases}
2x_1+x_2\le 10 & \text{(trigo)}\\
x_1+x_2\le 7 & \text{(molienda)}\\
x_1\le 4 & \text{(cuota integral)}\\
x_1,x_2\ge 0
\end{cases}$$

Resolviendo por Simplex —minimizando $-z$, igual que en el Ej. 1; se comprueba en las celdas de código— el **tableau óptimo** es (la celda objetivo muestra $z=-w=24$):

| Base | $x_1$ | $x_2$ | $s_1$ | $s_2$ | $s_3$ | RHS |
|:---:|:---:|:---:|:---:|:---:|:---:|:---:|
| **CR** | $0$ | $0$ | $1$ | $2$ | $0$ | $\mathbf{24}$ |
| $x_2$ | $0$ | $1$ | $-1$ | $2$ | $0$ | $4$ |
| $s_3$ | $0$ | $0$ | $-1$ | $1$ | $1$ | $1$ |
| $x_1$ | $1$ | $0$ | $1$ | $-1$ | $0$ | $3$ |

$$\boxed{x_1^*=3,\quad x_2^*=4,\quad z^*=24}$$

Los **precios sombra** se leen en la fila CR bajo las holguras: $y^\top=(1,\,2,\,0)$. La cuota de integral queda con holgura ($s_3=1$).

> Para la **dualidad** conviene mantener el primal en su forma natural de **máximo** (así el par primal-dual es el estándar máx $\leftrightarrow$ mín); la conversión a mínimo es solo la técnica de resolución del Simplex.

### (b) Problema dual

Primal de **maximización**, $3$ restricciones $\le$ y $2$ variables $\ge0$ → dual de **minimización**, $3$ variables $y_i\ge0$ y $2$ restricciones $\ge$ (una por columna del primal):

$$\min\; w=10y_1+7y_2+4y_3 \qquad \text{s.a.}\quad
\begin{cases}
2y_1+y_2+y_3\ge 4 & \text{(columna de }x_1)\\
y_1+y_2\ge 3 & \text{(columna de }x_2)\\
y_1,y_2,y_3\ge 0
\end{cases}$$

Interpretación: $y_i$ es el **valor imputado** al recurso $i$; el dual busca el mínimo costo de los recursos tal que cada harina quede valorada al menos en su margen.

### (c) Dualidad fuerte

De los precios sombra $y^*=(1,2,0)$:

$$w^*=10(1)+7(2)+4(0)=10+14+0=24.$$

$$\boxed{z^*=w^*=24\quad\checkmark\ \text{(dualidad fuerte)}}$$

Además $y^*$ es **factible** en el dual: $2(1)+2+0=4\ge4$ y $1+2=3\ge3$ (ambas activas).

### (d) Holguras complementarias e interpretación

Evaluamos el óptimo primal $x^*=(3,4)$:

| Restricción primal | Valor | ¿Activa? | Precio sombra |
|:---|:---:|:---:|:---:|
| Trigo: $2x_1+x_2\le10$ | $2(3)+4=10$ | sí | $y_1^*=1>0$ ✓ |
| Molienda: $x_1+x_2\le7$ | $3+4=7$ | sí | $y_2^*=2>0$ ✓ |
| Cuota integral: $x_1\le4$ | $3<4$ | **no** (holgura $1$) | $y_3^*=0$ ✓ |

Y por el lado de las variables: $x_1^*=3>0$ y $x_2^*=4>0$ ⇒ ambas restricciones **duales** están activas ($=4$ y $=3$). **Todas las condiciones de holgura complementaria se cumplen.**

**Interpretación de precios sombra:**
- $y_1^*=1$: una tonelada extra de **trigo** aumenta el margen en **1 mill\$**.
- $y_2^*=2$: una hora extra de **molienda** aumenta el margen en **2 mill\$** → es el **cuello de botella** más valioso.
- $y_3^*=0$: la **cuota de integral** no restringe (está holgada); ampliarla no mejora el óptimo.

### Comprobación con código

Resolvemos el primal y leemos los **precios sombra** (`.Pi` en Gurobi), verificando dualidad fuerte y holguras complementarias.

```python
from scipy.optimize import linprog
res = linprog(c=[-4, -3], A_ub=[[2,1],[1,1],[1,0]], b_ub=[10,7,4],
              bounds=[(0,None),(0,None)], method='highs')
print("scipy primal: x1=%.2f  x2=%.2f  z*=%.2f" % (res.x[0], res.x[1], -res.fun))

import gurobipy as gp
from gurobipy import GRB
m = gp.Model("molino"); m.Params.OutputFlag = 0
x1 = m.addVar(name="x1_integral"); x2 = m.addVar(name="x2_blanca")
m.setObjective(4*x1 + 3*x2, GRB.MAXIMIZE)
r1 = m.addConstr(2*x1 + x2 <= 10, "trigo")
r2 = m.addConstr(  x1 + x2 <=  7, "molienda")
r3 = m.addConstr(  x1      <=  4, "cuota_integral")
m.optimize()
print("gurobi primal: x1=%.2f  x2=%.2f  z*=%.2f" % (x1.X, x2.X, m.ObjVal))
print("Precios sombra (solucion dual):")
for r in (r1, r2, r3):
    print(f"   {r.ConstrName:15s} y={r.Pi:4.2f} | holgura={r.Slack:.2f}"
          + ("  <- holgura complementaria: holgada => y=0" if r.Slack > 1e-6 else ""))
w = 10*r1.Pi + 7*r2.Pi + 4*r3.Pi
print("Dual  w* = 10y1+7y2+4y3 = %.2f   =>  z* = w*  (dualidad fuerte)" % w)
```

|:---|:---|:---|
| **Convención** | El Simplex minimiza; un máximo se resuelve como $\min(-z)$ (costos con signo cambiado), y $z^*=-w^*$. El tableau queda idéntico. | fila CR inicial $=-$márgenes |

| **Simplex Matricial** | Mismas iteraciones con $B^{-1}$: $x_B=B^{-1}b$, columnas $B^{-1}A_j$, CR $=c_{NB}-c_BB^{-1}A_{NB}$, $w=c_BB^{-1}b$ | — |
| **El cruce** | $B^{-1}[A\mid I\mid b]=[B^{-1}A\mid B^{-1}\mid B^{-1}b]$: las columnas bajo las holguras **son** $B^{-1}$ | columnas $s_i$ = $B^{-1}$; RHS = $B^{-1}b$ |
| **Dualidad fuerte** | $z^*=w^*$ en el óptimo | valor de $z$ |
| **Precios sombra** | $y^\top=-c_BB^{-1}$; $y_i=\partial z^*/\partial b_i$ | fila CR bajo las holguras |
| **Holguras complementarias** | restricción holgada $\Rightarrow y_i=0$; $x_j>0\Rightarrow$ restricción dual activa | signos de holgura y CR |
