#### a) Convexidad (suficiencia)

- $e^x$: Función convexa

Como la función objetivo es la suma de funciones convexas, ella es convexa. La primera restricción se puede ver como el espacio definido por un plano, y la segunda como el cuadrante positivo del espacio (o dos restricciones con plano), por ello son también convexas.

En problemas convexos, cualquier punto que satisfaga KKT es **óptimo**, en este caso global pues se considera todo el especio de solución.

#### b) Formulación de Lagrangiano y condiciones KKT

$$ L(x,y,\mu_1, \mu_2, \mu_3) = e^x + y^2 + \mu_1 (x+y-1) + \mu_2 (-x + 1) + \mu_3 (-y + 1)$$

donde, $\mu_1, \mu_2, \mu_3 \ge 0$.

#### Condiciones KKT:

- Estacionariedad

$$ \frac{\partial L}{\partial x}: e^x + \mu_1 - \mu_2 = 0 $$
 
$$ \frac{\partial L}{\partial y}: 2y + \mu_1 - \mu_3 = 0 $$

$$ \mu_1 (x^*+y^*-1) = 0 $$

$$ \mu_2 (1-x^*) = 0 $$

$$ \mu_3 (1-y^*) = 0 $$

Las mismas restricciones, aunque por condición de Slater, en el óptimo la restricción debe ser $\lt 0$
