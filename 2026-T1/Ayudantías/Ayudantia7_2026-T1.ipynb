## 1. Teoría: ¿Por qué las restricciones ≥ complican el Simplex?

### 1.1 Repaso: Forma Estándar con Restricciones ≤

Cuando todas las restricciones son del tipo $\leq$, la conversión a forma estándar es directa: se agrega una **variable de holgura** $s_i \geq 0$ a cada restricción:

$$a_1 x_1 + a_2 x_2 \leq b \quad \xrightarrow{+s_i} \quad a_1 x_1 + a_2 x_2 + s_i = b$$

La colección de holguras forma automáticamente una **base inicial factible**: $B = I$ (identidad), con $x_B = b \geq 0$.

### 1.2 El Problema con Restricciones ≥

Cuando una restricción es del tipo $\geq$, se **resta** una variable de excedente (surplus) $e_i \geq 0$:

$$a_1 x_1 + a_2 x_2 \geq b \quad \xrightarrow{-e_i} \quad a_1 x_1 + a_2 x_2 - e_i = b$$

El problema: la columna de $e_i$ es $-e_i$ en la restricción, con coeficiente $-1$. Si usáramos $\{e_i\}$ como base, tendríamos $x_B = B^{-1}b$ donde la columna de $e_i$ tiene $-1$, lo que da **$x_B < 0$** para $b > 0$ — infactible.

> **En resumen:** las variables de excedente tienen signo negativo y no pueden formar una base factible por sí solas.

### 1.3 Solución: Variables Artificiales y el Método de la Gran M

Para cada restricción $\geq$ se agrega una **variable artificial** $a_i \geq 0$ que sí tiene coeficiente $+1$ (forma la identidad) y **permite arrancar** el algoritmo con una SBF:

$$a_1 x_1 + a_2 x_2 - e_i + a_i = b$$

Sin embargo, la variable artificial es **ficticia**: no pertenece al problema real. Para forzar que $a_i = 0$ en la solución óptima, se penaliza fuertemente su presencia en la función objetivo:

$$\text{Maximización:} \quad \max \; z = c^\top x - M \cdot a_i \qquad (M \gg 0)$$

Si $a_i > 0$ al final, el problema es **infactible** (no existe solución que satisfaga todas las restricciones).

### 1.4 Tabla Resumen de Variables

| Tipo de restricción | Variable que se agrega | Nombre | Coef. en obj. | ¿Forma base inicial? |
|:---:|:---:|:---:|:---:|:---:|
| $\leq$ | $+s_i$ | Holgura (slack) | $0$ | Sí ✓ |
| $\geq$ | $-e_i$ | Excedente (surplus) | $0$ | No ✗ |
| $\geq$ | $+a_i$ | Artificial | $-M$ (max) / $+M$ (min) | Sí ✓ (pero ficticia) |

### 1.5 Ajuste del Tableau Inicial

Al iniciar el tableau con artificiales en la base, la fila de costos reducidos (CR) **no** puede mostrar directamente $-c$ para los coeficientes de los artificiales — deben eliminarse de la fila objetivo mediante operaciones de fila, igual que en el método de Gauss-Jordan.

$$\text{CR}_{\text{nuevo}} = \text{CR}_{\text{inicial}} - M \cdot R_{a_i}$$

Esto garantiza que la columna de $a_i$ tenga cero en la fila CR (condición de variable básica).
