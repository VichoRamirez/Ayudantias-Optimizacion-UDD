### Ejercicio 22

Una empresa dedicada a la fabricación y comercialización de productos eléctricos debe planificar las labores de sus representantes comerciales en el país. La empresa trabaja con una cartera de $I$ clientes. Para cada cliente $i$ se conoce el número de visitas que se le debe realizar mensualmente $m_i$ y su volumen de compras mensuales, $d_i$ millones de pesos.

La empresa cuenta con $R$ representantes zonales. Cada uno de ellos, debe atender a un grupo de entre $A$ y $B$ clientes, cuyas compras totales no superen los $T$ millones de pesos al mes. El principal costo para la empresa en estas actividades corresponde a las visitas
que realizan los representantes a sus clientes. Se estima un costo $C_{ir}$ millones de pesos por cada visita que el representante r realiza al cliente i.

Formule un modelo de programación lineal entera que permita realizar una asignación de representantes a clientes de manera de cumplir con las condiciones mencionadas anteriormente minimizando el costo total de las visitas de los representantes a los clientes.

#### Variables

- $x_{ir}$: variable binaria que vale 1 si el cliente $i$ es asignado al representante zonal $r$; 0, si no.

#### Función objetivo.

El objetivo consiste en minimizar el costo total de las visitas que se calcula considerando el representante que atiende a cada cliente y el numero de visitas que debe realizar:

$$ min \quad \sum^X_{i=1} \sum^R_{r=1} m_i C_{ir} x_{ir} $$

#### Restricciones

a. Asignación de representante a cada cliente:

$$ \sum^R_{r=1} x_{ir} = 1 \quad  \quad , \forall i = 1,2,...,I$$

b. Número mínimo y máximo de clientes asignados a cada representante:

$$ A \le \sum^I_{i=1} x_{ir} \le B \quad \quad , \forall r = 1,2,...,R$$

c. Volumen de compras de clientes asignados a cada representante:

$$\sum^I_{i=1} d_i x_{ir} \le T \quad \quad , \forall r = 1,2,...,R$$

d. Naturaleza de las variables:

$$x_{ir} \in \left\{ 0,1 \right\} \quad \quad \forall i, \forall r$$

### Ejercicio 23

ENAP recibe petróleo crudo en sus refinerías, lo procesa y luego lo envía a sus plantas de distribución y almacenamiento (PDA) ubicadas en distintas zonas del país. La cantidad y ubicación de las refinerías actuales no pueden modificarse. Se desea reevaluar la ubicación y cantidad de las PDAs de ENAP, asi como el plan de procesamiento y distribución de petroleo para los siguientes $T$ periodos.

Existen varias ubicaciones potenciales donde podrían instalarse las PDA necesarias. Se sabe que $f_k$ es el costo de instalar una PDA en la ubicación $k \in K$ y que $c_{ik}$ es el costo de transportar un $m^3$ de petróleo procesado desde la refinería $i$ a la PDA $k$. Considere
que las PDA se pueden instalar sólo en el primer periodo de planificación, y que tienen dos tipos de capacidades máximas:

1. Cada PDA no puede recibir más de $q_k$ ($m^3$) de petróleo procesado en cada periodo.
2. No puede almacenar más de $h_k$ ($m^3$) desde un periodo a otro.

Almacenar petróleo de un periodo a otro no tiene costo. Considere que $s_{it}$ ($m3$) es la cantidad máxima de petróleo crudo que ENAP puede procesar en la refinería $i \in R$ el periodo $t \in T$ y que $p_i$ es el porcentaje del petróleo crudo que se recupera como petróleo procesado, el cual se envía a las PDA. Asuma que para este análisis se puede ignorar el costo de compra y procesamiento del petróleo crudo.

Finalmente, las ubicaciones de las potenciales PDA se agrupan en $U$ regiones, considere que el conjunto $V_r$ contienen a las ubicaciones potenciales de las PDA que pertenecen a la región $r \in U$. En cada periodo, las PDA de cada región deben disponer, en conjunto, de $d_{rt}$ $(m^3)$ de petróleo procesado para abastecer a sus clientes.

a. Formule un modelo de optimización lineal mixto que represente este problema y que minimice el costo de instalación de las PDA y de transporte.

b. Considere que a partir del periodo $n$ es posible aumentar la capacidad de inventario de algunas PDA. El aumento posible de capacidad de la PDA $k$ es de $p_k$ $(m^3)$ por un costo total de $b_k$. El aumento de la capacidad del inventario se mantiene desde el año en que se implementa hasta el último periodo inclusive. Muestre los cambios necesarios a su modelo de la parte anterior para considerar
esta nueva condición.

### Solución parte a

#### Variables de decisión

- $x_{it}$ cantidad de petróleo crudo a procesar en refinería $i$ en $t$.
- $y_{ikt}$ cantidad de petróleo procesado enviado desde refinería $i$ a la PDA $k$ en $t$.
- $z_{kt}$ cantidad de petróleo en inventario en PDA $k$ en $t$.
- $w_k$ (binaria) 1 si se instala una PDA en la ubicación k, 0 si no.

#### Función objetivo

$$ mín \quad \sum_{k \in K} f_k w_k + \sum_{i \in R} \sum_{k \in K} \sum_{t \in T} c_{ik} y_{ikt} $$

#### Restricciones

- Oferta de refinerías

$$ x_{it} \le s_{it} \quad \quad \forall i, t$$

- Relación producción envío en refinería

$$ p_i x_{it} − \sum_{k \in K} y_{ikt} = 0 \quad \quad \forall i, t $$

- Demanda regional

$$ \sum_{k \in V_r} y_{ikt} − \sum_{k \in V_r} z_{kt} \ge d_{rt} \quad \quad \forall r, t = 1$$

$$ \sum_{k \in V_r} y_{ikt} − \sum_{k \in V_r} z_{kt} + \sum_{k \in V_r} z_{k(t−1)} \ge d_{rt} \quad \quad \forall r, t > 1$$

- Restriccion capacidades máximas en PDA más relación entre variables, (tanto $y$ como $z$ deben relacionarse a la instalación de la PDA $w$)

$$ \sum_{i \in R} y_{ikt} \le q_k w_k \quad \quad \forall k, t $$

$$ z_{kt} \le h_k w_k \quad \quad \forall k, t$$

- Naturaleza de las variables o no negatividad

$$x_{it} \ge 0 \quad \quad \forall i, t$$
$$y_{ikt} \ge 0 \quad \quad \forall i, k, t$$
$$z_{kt} \ge 0 \quad \quad \forall k, t$$
$$w_k \in \{ 0, 1 \} \quad \quad \forall k$$

### Solución parte b

#### Nueva variable de decisión

- $v_{kt}$ (binaria) 1 si se aumenta la capacidad de inventario del PDA $k$ en $t$, 0 si no.

A la función objetivo se debe agregar el costo del aumento de la capacidad:

$$ \cdots + \sum_{k \in K} \sum_{t \in T} b_k v_{kt} $$

Se modifica la restricción de capacidad de inventario:

$$z_{kt} \ge h_k + \sum^t_{t^′=n} o_k v_{kt^′} \quad \quad \forall k, t$$

Se debe restringir que el aumento se puede realizar una sola vez:

$$\sum_{t \in T} v_{kt} \le 1 \quad \quad \forall k$$
