Metodo Simplex - El problema Dual

La teoria de dualidad, la puede ver en: https://www.youtube.com/watch?v=8wP3zoXXWlA

Hasta ahora hemos resuelto el problema

$$ mín \quad c^T \cdot x $$

sujeto a,

$$ A \cdot x \le b $$

$$ x_i \ge 0 $$

Para pasar a forma estándar, agregamos variables *slack* (holgura), de modo de modificar el problema y tener una solución inicial factible.

#### Prueba de optimalidad

Esta prueba la podemos aplicar sobre una solución inicial factible, pero hay casos en que al agregar las holguras no tenemos solución inicial factible.

$$ c^T - c_B^T \cdot B^{-1} A $$

Recordar que si algún coeficiente es negativo, la solución no es óptima. Para solucionar esto resolvemos el problema dual.

#### Problema Primal vs Dual

El problema primal es el que hasta hoy hemos visto, tiene el problema dual asociado.

$$ max \quad b^T \cdot y $$

sujeto a,

$$ A^T \cdot y \ge c $$

$$ y_i \ge 0 $$

La resolución del problema primal lo que hace, es partir desde una solución factible, realizar la prueba de optimalidad e iterar, si se itera se escoge el minimo entre los cuocientes factibles de la restriccion y columna de la base entrante, como es el minimo lo que hace es mantener el vector $b$ positivo, y de esa manera la solución se mantiene factible.

En cambio el problema dual, comienza de una solución no factible, pero óptima. Al igual que el problema anterior, se puede iterar y probar en ese caso factibilidad, aquí la iteración es similar al método simplex anterior, pero se mantiene que sea la solución sea óptima, hasta encontrar la factibilidad.

Primero veamos el algoritmo **simplex dual** desde la perspectiva tableau, con el ejemplo de la clase anterior.

#### Iteración

Se elige el coeficiente más negativo y este entra a la base (sale min entre lado entre cuociente del lado derecho y columna).

| V.Básica | Ecuación | $\vec{x}$ | LD |
|:-:|:-:|:-:|:-:|
| z | F.Obj. | $c-c_B B^{-1}A$ | $c_B B^{-1}b$ |
| $\vec{x_B}$ | Rest. | $B^{-1} A$ | $B^{-1}b$ |


### Ejercicio 18

$$min \quad 25x1 + 20x2$$

s.a.

$$4x1 + 3x2 \ge 250$$
$$3x1 + 4x2 \ge 270$$
$$2x1 + 5x2 \ge 300$$
$$x1; x2 \ge 0$$

Para este ejercicio agregamos holguras negativas, si hacemos el tableau.

En este caso, el tableau nos dice que $s_1 = -250$, $s_2 = -270$, $s_3 = -300$, pero las variables tiene condición de no negatividad, lo que implica que la solución es no factible. Al ser la fila $z$ positivos, el algoritmo anterior dice que es óptimo (fijese que el minimo es 0, y como no hay negatividad de las variables, no puede ser nunca menor que 0). Modifiquemos entonces el tableau y modificamos el algoritmo.
