Metodo Simplex - El problema Dual

#### Primal:

$$ mín \quad c^T \cdot x $$

sujeto a,

$$ A \cdot x \le b $$

$$ x_i \ge 0 $$

#### Dual

$$ max \quad b^T \cdot y $$

sujeto a,

$$ A^T \cdot y \ge c $$

$$ y_i \ge 0 $$

#### Teorema de las holguras complementarias

Se utiliza para obtener la solución del problema complementario, teniendo la solución del otro problema, consiste que si la restricción es activa (es igualdad, holgura = 0), la variable asociada en el dual no lo será (distinto de 0), y al reves, es decir si la restricción no es activa, la variable asociada en el dual es 0.

Equivale a **Condición de holgura complementaria KKT**: $\mu_i g_i(x^*) = 0$
