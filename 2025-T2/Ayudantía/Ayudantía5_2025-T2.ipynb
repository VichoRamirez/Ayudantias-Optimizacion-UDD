```python
import math
import gurobipy as gp
from gurobipy import GRB

f = open("sudoku.txt")

grid = f.read().split()

n = len(grid[0])
s = int(math.sqrt(n))
print(grid)

model = gp.Model("Sudoku")

x = model.addVars(n, n, n, name = "Numero", lb = 0, vtype = GRB.BINARY)

for i in range(n):
    for j in range(n):
        if grid[i][j] != ".":
            v = int(grid[i][j]) - 1
            x[i, j, v].LB = 1

for columna in range(n):
    for numero in range(n):
        model.addConstr(gp.quicksum(x[fila, columna, numero] for fila in range(n))==1 )

model.addConstrs(
    gp.quicksum(x[fila,columna,numero] for columna in range(n))==1 
    for numero in range(n) for fila in range(n))

model.addConstrs(
    gp.quicksum(x[fila,columna,numero] for numero in range(n))==1 
    for columna in range(n) for fila in range(n))

model.addConstrs(gp.quicksum(x[i, j, v] for i in range(i0*s, (i0+1)*s) 
                             for j in range(j0*s, (j0+1)*s)) == 1for v in range(n)for i0 in range(s)for j0 in range(s))

model.optimize()
print("Constrains: ", model.NumConstrs)

model.write("sudoku.lp")

# Mostrar el grid del Sudoku
print("\nSolución del Sudoku:\n")

solution = model.getAttr("X", x)
for i in range(n):
    row = []
    for j in range(n):
        for v in range(n):
            if solution[i, j, v] > 0.5:
                row.append(str(v + 1))  # Convertir el valor a string

    # Imprimir la fila con separadores de cuadrantes
    print("| "+" | ".join([" ".join(row[j:j+s]) for j in range(0, n, s)]) + " |")  # Separar por subgrids

    # Agregar una línea horizontal cada s filas (excepto al final)
    if (i + 1) % s == 0 and i + 1 < n:
        print("-" * ((n + s)*2+1))  # Línea horizontal con separadores
```
